# AI_EMBED

```
%%sql -r embeddings_result
SELECT
    date_created,
    transcript,
    AI_EMBED('snowflake-arctic-embed-l-v2.0', transcript) AS transcript_embedding
FROM call_transcripts
LIMIT 5;
```

# AI_SIMILARITY

```
SELECT
AI_SIMILARITY(
'broken zipper',
'the zipper snapped off'
) AS similarity_score;
```

```
SELECT
AI_SIMILARITY(
'broken zipper',
'my order has not arrived yet'
) AS similarity_score;
```

```
%%sql -r dataframe_3
SELECT
date_created,
transcript,
AI_SIMILARITY(
'broken zipper on the Ridgeline jacket',
transcript
) AS relevance_score
FROM call_transcripts 
ORDER BY relevance_score DESC
LIMIT 10;
```
